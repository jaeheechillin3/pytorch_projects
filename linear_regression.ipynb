```python
'''
1. preparing data
2. building the model and the loss function
3. training the model for a single parameter
4. training the model for two parameters
'''
import torch
import numpy as np
import matplotlib.pyplot as plt

# creating synthetic data - a linear function from -5 to 5
X = torch.arange(-5, 5, 0.1).view(-1, 1)
func = -5 * X

plt.plot(X.numpy(), func.numpy(), 'r', label='func')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid('True', color='y')
plt.show()
```

```python
# adding Gaussian noise
Y = func + 0.4 * torch.randn(X.size())

plt.plot(X.numpy(), Y.numpy(), 'b+', label='Y')
plt.plot(X.numpy(), func.numpy(), 'r', label='func')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid('True', color='y')
plt.show()
```
